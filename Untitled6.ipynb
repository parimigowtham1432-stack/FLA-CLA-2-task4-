````python

# =====================================================================
# SRM INSTITUTE OF SCIENCE AND TECHNOLOGY, TIRUCHIRAPPALLI
# 21CSC301T - FORMAL LANGUAGE AND AUTOMATA | CLA-2 | TASK 4
# CASE STUDY: COMPILER BUG DETECTION
#
# GROUP MEMBERS:
# 1. PARIMI GOWTHAM - RA2411026050033
# 2. DANDE DURGA PRASAD - RA2411026050066
# 3. M. SHANMUKHA CHOWDHARY - RA2411026050064
#
# Department: AIML | Semester: V | Academic Year: 2026-2027
# Faculty: Dr. C. Gunasundari
#
# ONE-CELL GOOGLE COLAB IMPLEMENTATION
# =====================================================================

import re
import random
import pandas as pd
import matplotlib.pyplot as plt
from dataclasses import dataclass
from IPython.display import display, Markdown

pd.set_option("display.max_colwidth", 120)
random.seed(42)


def H(text):
    display(Markdown("# " + text))


def M(text):
    display(Markdown(text))


H("TASK 4: COMPILER BUG DETECTION")

M("""
**Case study:** Detecting semantic defects in a compiler by comparing a
source-language reference interpreter with compiled bytecode execution.

**Restricted language:** MiniLang supports integer variables, assignments,
print, arithmetic expressions, if/else, and while loops.

The compiler translates this subset into a small virtual-machine instruction
set. We inject a faulty optimization so that the testing framework has a
known defect to find.

**Assumptions:** deterministic execution; integer arithmetic; no external
I/O other than print; finite test inputs; a configurable instruction-step
limit prevents the notebook from hanging. A timeout is reported as
UNKNOWN / STEP LIMIT, not as proof of an infinite loop.
""")


# =====================================================================
# 1. FORMAL LANGUAGE: TOKENIZER, GRAMMAR AND AST
# =====================================================================

M("""
## 1. Source language and grammar

The educational MiniLang grammar is:

```text
Program   -> Stmt*
Stmt      -> id = Expr ;
           | print ( Expr ) ;
           | if ( Bool ) { Stmt* } else { Stmt* }
           | while ( Bool ) { Stmt* }

Bool      -> Expr RelOp Expr
RelOp     -> < | <= | > | >= | == | !=

Expr      -> Term ((+ | -) Term)*
Term      -> Factor ((* | / | %) Factor)*
Factor    -> INT | id | ( Expr )
```

This is a restricted language, not full Python, C, or Java.
""")

TOKEN_RE = re.compile(
    r"\s*(?:(\d+)|([A-Za-z_][A-Za-z_0-9]*)|(==|!=|<=|>=|[+\-*/%<>=(){};]))"
)


def tokenize(source):
    tokens, pos = [], 0

    while pos < len(source):

        if source[pos:].strip() == "":
            break

        match = TOKEN_RE.match(source, pos)

        if not match:
            raise SyntaxError(
                f"Unexpected character {source[pos]!r} at position {pos}"
            )

        number, ident, symbol = match.groups()

        if number is not None:
            tokens.append(("INT", int(number)))

        elif ident is not None:

            if ident in ("if", "else", "while", "print"):
                tokens.append((ident, ident))
            else:
                tokens.append(("ID", ident))

        else:
            tokens.append((symbol, symbol))

        pos = match.end()

    tokens.append(("EOF", None))

    return tokens


@dataclass
class Expr:
    kind: str
    value: object = None
    left: object = None
    right: object = None


@dataclass
class Stmt:
    kind: str
    name: object = None
    expr: object = None
    left: object = None
    right: object = None
    body: object = None
    other: object = None


class MiniLangParser:

    def __init__(self, source):

        self.tokens = tokenize(source)
        self.i = 0
        self.trace = []

    def cur(self):
        return self.tokens[self.i]

    def eat(self, kind=None):

        tok = self.cur()

        if kind is not None and tok[0] != kind:
            raise SyntaxError(
                f"Expected {kind}, found {tok[0]}"
            )

        self.i += 1

        self.trace.append(
            f"Consumed {tok[0]} {tok[1]}"
        )

        return tok

    def parse(self):

        statements = []

        while self.cur()[0] != "EOF":
            statements.append(self.statement())

        return statements

    def block(self):

        self.eat("{")

        items = []

        while self.cur()[0] not in ("}", "EOF"):
            items.append(self.statement())

        self.eat("}")

        return items

    def statement(self):

        kind = self.cur()[0]

        if kind == "ID":

            name = self.eat("ID")[1]

            self.eat("=")

            value = self.expression()

            self.eat(";")

            return Stmt(
                "assign",
                name=name,
                expr=value
            )

        if kind == "print":

            self.eat("print")
            self.eat("(")

            value = self.expression()

            self.eat(")")
            self.eat(";")

            return Stmt(
                "print",
                expr=value
            )

        if kind == "if":

            self.eat("if")
            self.eat("(")

            cond = self.boolean()

            self.eat(")")

            yes = self.block()

            self.eat("else")

            no = self.block()

            return Stmt(
                "if",
                expr=cond,
                body=yes,
                other=no
            )

        if kind == "while":

            self.eat("while")
            self.eat("(")

            cond = self.boolean()

            self.eat(")")

            body = self.block()

            return Stmt(
                "while",
                expr=cond,
                body=body
            )

        raise SyntaxError(
            f"Unexpected statement token {kind}"
        )

    def boolean(self):

        left = self.expression()

        op = self.cur()[0]

        if op not in ("<", "<=", ">", ">=", "==", "!="):
            raise SyntaxError("Expected relational operator")

        self.eat()

        right = self.expression()

        return Expr(
            "compare",
            op,
            left,
            right
        )

    def expression(self):

        node = self.term()

        while self.cur()[0] in ("+", "-"):

            op = self.eat()[0]

            node = Expr(
                "binary",
                op,
                node,
                self.term()
            )

        return node

    def term(self):

        node = self.factor()

        while self.cur()[0] in ("*", "/", "%"):

            op = self.eat()[0]

            node = Expr(
                "binary",
                op,
                node,
                self.factor()
            )

        return node

    def factor(self):

        tok = self.cur()

        if tok[0] == "INT":

            self.eat("INT")

            return Expr("int", tok[1])

        if tok[0] == "ID":

            self.eat("ID")

            return Expr("var", tok[1])

        if tok[0] == "(":

            self.eat("(")

            node = self.expression()

            self.eat(")")

            return node

        raise SyntaxError(
            f"Expected integer, identifier or parenthesis; got {tok[0]}"
        )


def parse_program(source):

    parser = MiniLangParser(source)

    result = parser.parse()

    return result, parser.trace


# =====================================================================
# 2. REFERENCE INTERPRETER
# =====================================================================

class MiniLangRuntimeError(Exception):
    pass


def eval_expr(node, env):

    if node.kind == "int":
        return node.value

    if node.kind == "var":

        if node.value not in env:
            raise MiniLangRuntimeError(
                f"Undefined variable: {node.value}"
            )

        return env[node.value]

    if node.kind == "binary":

        a = eval_expr(node.left, env)
        b = eval_expr(node.right, env)

        if node.value == "+":
            return a + b

        if node.value == "-":
            return a - b

        if node.value == "*":
            return a * b

        if node.value == "/":

            if b == 0:
                raise MiniLangRuntimeError("Division by zero")

            return a // b

        if node.value == "%":

            if b == 0:
                raise MiniLangRuntimeError("Modulo by zero")

            return a % b

    if node.kind == "compare":

        a = eval_expr(node.left, env)
        b = eval_expr(node.right, env)

        return {
            "<": a < b,
            "<=": a <= b,
            ">": a > b,
            ">=": a >= b,
            "==": a == b,
            "!=": a != b
        }[node.value]

    raise MiniLangRuntimeError("Unknown expression node")


def run_reference(statements, max_steps=10000):

    env = {}
    output = []
    steps = 0

    def run_block(block):

        nonlocal steps

        for stmt in block:

            steps += 1

            if steps > max_steps:
                raise TimeoutError(
                    "Reference execution step limit reached"
                )

            if stmt.kind == "assign":

                env[stmt.name] = eval_expr(
                    stmt.expr,
                    env
                )

            elif stmt.kind == "print":

                output.append(
                    eval_expr(stmt.expr, env)
                )

            elif stmt.kind == "if":

                branch = (
                    stmt.body
                    if eval_expr(stmt.expr, env)
                    else stmt.other
                )

                run_block(branch)

            elif stmt.kind == "while":

                while eval_expr(stmt.expr, env):

                    steps += 1

                    if steps > max_steps:
                        raise TimeoutError(
                            "Reference execution step limit reached"
                        )

                    run_block(stmt.body)

    run_block(statements)

    return tuple(output), dict(env), steps


# =====================================================================
# 3. COMPILER TO STACK VIRTUAL MACHINE
# =====================================================================

def compile_expression(node, instructions, faulty=False):

    if node.kind == "int":

        instructions.append(
            ("PUSH", node.value)
        )

    elif node.kind == "var":

        instructions.append(
            ("LOAD", node.value)
        )

    elif node.kind == "binary":

        compile_expression(
            node.left,
            instructions,
            faulty
        )

        compile_expression(
            node.right,
            instructions,
            faulty
        )

        op = node.value

        instructions.append(
            ("BIN", op)
        )

    elif node.kind == "compare":

        compile_expression(
            node.left,
            instructions,
            faulty
        )

        compile_expression(
            node.right,
            instructions,
            faulty
        )

        instructions.append(
            ("CMP", node.value)
        )

    else:

        raise ValueError("Unknown expression")


def compile_program(statements, faulty=False):

    code = []

    def emit_block(block):

        for stmt in block:

            if stmt.kind == "assign":

                compile_expression(
                    stmt.expr,
                    code,
                    faulty
                )

                code.append(
                    ("STORE", stmt.name)
                )

            elif stmt.kind == "print":

                compile_expression(
                    stmt.expr,
                    code,
                    faulty
                )

                code.append(
                    ("PRINT", None)
                )

            elif stmt.kind == "if":

                compile_expression(
                    stmt.expr,
                    code,
                    faulty
                )

                jf_index = len(code)

                code.append(
                    ("JUMP_IF_FALSE", None)
                )

                emit_block(stmt.body)

                jmp_index = len(code)

                code.append(
                    ("JUMP", None)
                )

                else_start = len(code)

                emit_block(stmt.other)

                end = len(code)

                code[jf_index] = (
                    "JUMP_IF_FALSE",
                    else_start
                )

                code[jmp_index] = (
                    "JUMP",
                    end
                )

            elif stmt.kind == "while":

                start = len(code)

                compile_expression(
                    stmt.expr,
                    code,
                    faulty
                )

                exit_index = len(code)

                code.append(
                    ("JUMP_IF_FALSE", None)
                )

                emit_block(stmt.body)

                code.append(
                    ("JUMP", start)
                )

                code[exit_index] = (
                    "JUMP_IF_FALSE",
                    len(code)
                )

    emit_block(statements)

    code.append(("HALT", None))

    return code


# =====================================================================
# 4. BYTECODE VIRTUAL MACHINE
# =====================================================================

def run_bytecode(code, max_steps=10000):

    env = {}
    stack = []
    output = []

    pc = 0
    steps = 0
    trace = []

    while pc < len(code):

        steps += 1

        if steps > max_steps:
            raise TimeoutError(
                "Virtual-machine step limit reached"
            )

        op, arg = code[pc]

        old_pc = pc

        pc += 1

        if op == "PUSH":

            stack.append(arg)

        elif op == "LOAD":

            if arg not in env:
                raise MiniLangRuntimeError(
                    f"Undefined variable: {arg}"
                )

            stack.append(env[arg])

        elif op == "STORE":

            env[arg] = stack.pop()

        elif op == "BIN":

            b = stack.pop()
            a = stack.pop()

            if arg == "+":
                stack.append(a + b)

            elif arg == "-":
                stack.append(a - b)

            elif arg == "*":
                stack.append(a * b)

            elif arg == "/":

                if b == 0:
                    raise MiniLangRuntimeError("Division by zero")

                stack.append(a // b)

            elif arg == "%":

                if b == 0:
                    raise MiniLangRuntimeError("Modulo by zero")

                stack.append(a % b)

        elif op == "CMP":

            b = stack.pop()
            a = stack.pop()

            stack.append({
                "<": a < b,
                "<=": a <= b,
                ">": a > b,
                ">=": a >= b,
                "==": a == b,
                "!=": a != b
            }[arg])

        elif op == "JUMP_IF_FALSE":

            condition = stack.pop()

            if not condition:
                pc = arg

        elif op == "JUMP":

            pc = arg

        elif op == "PRINT":

            output.append(stack.pop())

        elif op == "HALT":

            trace.append(
                f"pc={old_pc}: HALT"
            )

            break

        trace.append(
            f"pc={old_pc}: {op} {arg}; env={env}; stack={stack}"
        )

    return tuple(output), dict(env), steps, trace


# =====================================================================
# 5. COMPILER OPTIMIZATION AND INJECTED BUG
# =====================================================================

M("""
## 2. Deliberately faulty compiler optimization

Correct constant-folding behavior preserves the meaning of expressions.

For demonstration, the faulty optimization applies this incorrect rule:

`0 - x -> x`

This changes program meaning. The defect is intentionally injected to check
whether our differential-testing framework can detect it. The reference
interpreter does not use this optimization and serves as the comparison oracle.
""")


def optimize_expression(node, inject_bug=False):

    if node is None:
        return None

    if node.kind in ("int", "var"):
        return node

    left = optimize_expression(
        node.left,
        inject_bug
    )

    right = optimize_expression(
        node.right,
        inject_bug
    )

    if node.kind == "binary":

        # Faulty transformation: 0 - x becomes x instead of -x.
        if (
            inject_bug
            and node.value == "-"
            and left.kind == "int"
            and left.value == 0
        ):

            return right

        # Correct constant folding for literal operands.
        if left.kind == "int" and right.kind == "int":

            a = left.value
            b = right.value
            op = node.value

            if op == "+":
                return Expr("int", a + b)

            if op == "-":
                return Expr("int", a - b)

            if op == "*":
                return Expr("int", a * b)

            if op == "/" and b != 0:
                return Expr("int", a // b)

            if op == "%" and b != 0:
                return Expr("int", a % b)

    return Expr(
        node.kind,
        node.value,
        left,
        right
    )


def optimize_statements(statements, faulty=False):

    result = []

    for s in statements:

        if s.kind in ("assign", "print"):

            result.append(
                Stmt(
                    s.kind,
                    name=s.name,
                    expr=optimize_expression(
                        s.expr,
                        faulty
                    )
                )
            )

        elif s.kind == "if":

            result.append(
                Stmt(
                    "if",
                    expr=optimize_expression(
                        s.expr,
                        faulty
                    ),
                    body=optimize_statements(
                        s.body,
                        faulty
                    ),
                    other=optimize_statements(
                        s.other,
                        faulty
                    )
                )
            )

        elif s.kind == "while":

            result.append(
                Stmt(
                    "while",
                    expr=optimize_expression(
                        s.expr,
                        faulty
                    ),
                    body=optimize_statements(
                        s.body,
                        faulty
                    )
                )
            )

    return result


# =====================================================================
# 6. DIFFERENTIAL TESTING FRAMEWORK
# =====================================================================

def execute_case(name, source, faulty=False, max_steps=10000):

    try:

        ast_nodes, parse_trace = parse_program(source)

        ref_output, ref_env, ref_steps = run_reference(
            ast_nodes,
            max_steps
        )

        optimized_ast = optimize_statements(
            ast_nodes,
            faulty=faulty
        )

        bytecode = compile_program(
            optimized_ast,
            faulty=faulty
        )

        vm_output, vm_env, vm_steps, vm_trace = run_bytecode(
            bytecode,
            max_steps
        )

        same = (
            ref_output == vm_output
            and ref_env == vm_env
        )

        return {
            "name": name,
            "source": source,
            "reference_output": str(ref_output),
            "compiled_output": str(vm_output),
            "reference_env": str(ref_env),
            "compiled_env": str(vm_env),
            "status": "PASS" if same else "BUG DETECTED",
            "parse_trace": parse_trace,
            "bytecode": bytecode,
            "vm_trace": vm_trace,
            "error": ""
        }

    except TimeoutError as e:

        return {
            "name": name,
            "source": source,
            "reference_output": "UNKNOWN",
            "compiled_output": "UNKNOWN",
            "reference_env": "",
            "compiled_env": "",
            "status": "UNKNOWN / STEP LIMIT",
            "parse_trace": [],
            "bytecode": [],
            "vm_trace": [],
            "error": str(e)
        }

    except Exception as e:

        return {
            "name": name,
            "source": source,
            "reference_output": "ERROR",
            "compiled_output": "ERROR",
            "reference_env": "",
            "compiled_env": "",
            "status": "ERROR",
            "parse_trace": [],
            "bytecode": [],
            "vm_trace": [],
            "error": str(e)
        }


# 15 curated benchmark programs.
BENCHMARK = [

    ("simple_add", "x = 4 + 3; print(x);"),

    ("simple_subtract", "x = 9 - 2; print(x);"),

    ("multiplication", "x = 6 * 7; print(x);"),

    ("precedence", "x = 2 + 3 * 4; print(x);"),

    ("parentheses", "x = (2 + 3) * 4; print(x);"),

    ("if_true",
     "x = 5; if (x > 2) { print(1); } else { print(0); }"),

    ("if_false",
     "x = 1; if (x > 2) { print(1); } else { print(0); }"),

    ("countdown",
     "x = 4; while (x > 0) { x = x - 1; } print(x);"),

    ("accumulator",
     "x = 0; i = 1; while (i <= 4) { x = x + i; i = i + 1; } print(x);"),

    ("modulo", "x = 17 % 5; print(x);"),

    ("zero_minus_variable",
     "x = 7; y = 0 - x; print(y);"),

    ("zero_minus_expression",
     "x = 4; y = 0 - (x + 2); print(y);"),

    ("negative_result", "x = 3 - 8; print(x);"),

    ("nested_if",
     "x = 3; if (x > 0) { if (x < 5) { print(10); } else { print(20); } } else { print(0); }"),

    ("division_zero",
     "x = 4; y = 0; z = x / y; print(z);")
]


correct_results = [
    execute_case(n, s, faulty=False)
    for n, s in BENCHMARK
]

faulty_results = [
    execute_case(n, s, faulty=True)
    for n, s in BENCHMARK
]


H("3. Benchmark: correct compiler")

correct_df = pd.DataFrame([{

    "Program": r["name"],

    "Source": r["source"],

    "Reference output": r["reference_output"],

    "Compiled output": r["compiled_output"],

    "Result": r["status"],

    "Details": r["error"]

} for r in correct_results])

display(correct_df)


H("4. Benchmark: compiler with injected optimization bug")

faulty_df = pd.DataFrame([{

    "Program": r["name"],

    "Source": r["source"],

    "Reference output": r["reference_output"],

    "Faulty output": r["compiled_output"],

    "Result": r["status"],

    "Details": r["error"]

} for r in faulty_results])

display(faulty_df)


# =====================================================================
# 7. WORKED EXAMPLE WITH COMPILER TRACE
# =====================================================================

H("5. Worked example: detecting the faulty optimization")

worked_source = "x = 7; y = 0 - x; print(y);"

worked_good = execute_case(
    "worked_example",
    worked_source,
    faulty=False
)

worked_bad = execute_case(
    "worked_example",
    worked_source,
    faulty=True
)


M(f"""
**Source program**

```text
{worked_source}
```

The reference interpreter calculates `0 - 7 = -7`. The correct compiler
also returns `-7`. The faulty optimization rewrites `0 - x` as `x`, so the
compiled program returns `7`. The differential tester reports
**{worked_bad['status']}**.

| Execution path | Output |
|---|---:|
| Reference interpreter | {worked_good['reference_output']} |
| Correct compiler | {worked_good['compiled_output']} |
| Faulty compiler | {worked_bad['compiled_output']} |
""")


print("Faulty compiler bytecode:")

for instruction in worked_bad["bytecode"]:
    print(" ", instruction)


print("\nFaulty virtual-machine trace:")

for line in worked_bad["vm_trace"]:
    print(" ", line)


# =====================================================================
# 8. FORMAL PROBLEM AND COMPUTABILITY
# =====================================================================

H("6. Formal problem definition and computability analysis")

M("""
Let:

- `P` be a source program.
- `C` be a compiler.
- `x` be an input.
- `⟦P⟧(x)` be the observable behavior of the source program.
- `⟦C(P)⟧(x)` be the observable behavior of the compiled program.

A compiler defect is witnessed by a program and input for which:

`⟦P⟧(x) != ⟦C(P)⟧(x)`

Universal compiler correctness asks whether:

`For every program P and every valid input x, ⟦P⟧(x) = ⟦C(P)⟧(x).`

For unrestricted programming languages, general semantic equivalence and
universal correctness questions are undecidable in general. Program
equivalence is a non-trivial semantic property, so Rice's theorem explains
why no algorithm can decide all such cases for arbitrary programs.

The Halting Problem is the central computability limit:

`HALT_TM = { <M,w> : Turing machine M halts on input w }`

No Turing machine decides membership in this language for every encoded
machine and input. A hypothetical general program-behavior verifier could
be used to decide termination/equivalence questions, contradicting this
result.

**What our implementation can do:** it decides whether the source and
compiled outputs match for each finite test that completes within the
configured step limit. It can expose a counterexample. Passing the tests
does not prove correctness for every possible program.
""")


# =====================================================================
# 9. DIAGRAMS
# =====================================================================

H("7. Diagram 1: real-world problem to computational model")

fig, ax = plt.subplots(figsize=(12, 5))

ax.set_xlim(0, 12)
ax.set_ylim(0, 5)
ax.axis("off")


def draw_box(ax, x, y, w, h, text, fontsize=9):

    ax.add_patch(
        plt.Rectangle(
            (x, y),
            w,
            h,
            fill=False,
            linewidth=1.8
        )
    )

    ax.text(
        x + w / 2,
        y + h / 2,
        text,
        ha="center",
        va="center",
        fontsize=fontsize
    )


draw_box(
    ax, 0.3, 3.25, 2.2, 1.0,
    "Source program P\n+ input x"
)

draw_box(
    ax, 3.1, 3.25, 2.2, 1.0,
    "Compiler C\ntranslation"
)

draw_box(
    ax, 5.9, 3.25, 2.2, 1.0,
    "Compiled program\nC(P)"
)

draw_box(
    ax, 8.7, 3.25, 2.8, 1.0,
    "Execute and observe\ncompiled behavior"
)


for x1, x2 in [
    (2.5, 3.1),
    (5.3, 5.9),
    (8.1, 8.7)
]:

    ax.annotate(
        "",
        xy=(x2, 3.75),
        xytext=(x1, 3.75),
        arrowprops=dict(
            arrowstyle="->",
            lw=1.5
        )
    )


draw_box(
    ax, 1.0, 0.9, 3.0, 1.0,
    "Reference interpreter\n⟦P⟧(x)"
)

draw_box(
    ax, 5.0, 0.9, 3.0, 1.0,
    "Differential comparison\nsource vs compiled"
)

draw_box(
    ax, 9.0, 0.9, 2.0, 1.0,
    "Mismatch?\nBug evidence"
)


ax.annotate(
    "",
    xy=(2.5, 1.9),
    xytext=(1.4, 3.25),
    arrowprops=dict(arrowstyle="->", lw=1.4)
)

ax.annotate(
    "",
    xy=(5.0, 1.4),
    xytext=(4.0, 1.4),
    arrowprops=dict(arrowstyle="->", lw=1.4)
)

ax.annotate(
    "",
    xy=(9.0, 1.4),
    xytext=(8.0, 1.4),
    arrowprops=dict(arrowstyle="->", lw=1.4)
)

ax.annotate(
    "",
    xy=(6.5, 1.9),
    xytext=(10.0, 3.25),
    arrowprops=dict(arrowstyle="->", lw=1.4)
)

ax.set_title(
    "Compiler Bug Detection: Real-world to Computational Model",
    fontsize=13,
    weight="bold"
)

plt.show()


# ---------------------------------------------------------------------

H("8. Diagram 2: compiler testing architecture")

fig, ax = plt.subplots(figsize=(11, 5))

ax.set_xlim(0, 11)
ax.set_ylim(0, 5)
ax.axis("off")


architecture = [

    (0.4, 3.3, 2.0, 0.9, "MiniLang source"),

    (3.0, 3.3, 2.0, 0.9, "Parser + AST"),

    (5.6, 3.3, 2.0, 0.9, "Optimization pass"),

    (8.2, 3.3, 2.2, 0.9, "Bytecode compiler"),

    (8.2, 1.0, 2.2, 0.9, "Virtual machine"),

    (4.4, 1.0, 2.6, 0.9, "Differential tester"),

    (0.7, 1.0, 2.5, 0.9, "Reference interpreter")

]


for x, y, w, h, t in architecture:

    draw_box(ax, x, y, w, h, t)


for x1, x2 in [
    (2.4, 3.0),
    (5.0, 5.6),
    (7.6, 8.2)
]:

    ax.annotate(
        "",
        xy=(x2, 3.75),
        xytext=(x1, 3.75),
        arrowprops=dict(arrowstyle="->", lw=1.5)
    )


ax.annotate(
    "",
    xy=(9.3, 1.9),
    xytext=(9.3, 3.3),
    arrowprops=dict(arrowstyle="->", lw=1.5)
)

ax.annotate(
    "",
    xy=(7.0, 1.45),
    xytext=(8.2, 1.45),
    arrowprops=dict(arrowstyle="->", lw=1.5)
)

ax.annotate(
    "",
    xy=(4.4, 1.45),
    xytext=(3.2, 1.45),
    arrowprops=dict(arrowstyle="->", lw=1.5)
)

ax.set_title(
    "Restricted Compiler-Testing System Architecture",
    fontsize=13,
    weight="bold"
)

plt.show()


# ---------------------------------------------------------------------

H("9. Diagram 3: Halting Problem and undecidability")

fig, ax = plt.subplots(figsize=(10, 4.5))

ax.set_xlim(0, 10)
ax.set_ylim(0, 4.5)
ax.axis("off")


draw_box(
    ax, 0.4, 1.7, 2.2, 1.0,
    "Encoded Turing machine M\nand input w"
)

draw_box(
    ax, 3.8, 1.7, 2.5, 1.0,
    "Hypothetical universal\nHALT decider"
)

draw_box(
    ax, 7.5, 2.8, 2.0, 0.8,
    "HALTS"
)

draw_box(
    ax, 7.5, 0.8, 2.0, 0.8,
    "LOOPS"
)


ax.annotate(
    "",
    xy=(3.8, 2.2),
    xytext=(2.6, 2.2),
    arrowprops=dict(arrowstyle="->", lw=1.5)
)

ax.annotate(
    "",
    xy=(7.5, 3.2),
    xytext=(6.3, 2.3),
    arrowprops=dict(arrowstyle="->", lw=1.5)
)

ax.annotate(
    "",
    xy=(7.5, 1.2),
    xytext=(6.3, 2.1),
    arrowprops=dict(arrowstyle="->", lw=1.5)
)

ax.text(
    5, 4.0,
    "No such decider works for every M and w.",
    ha="center",
    weight="bold"
)

ax.set_title(
    "Theoretical Limit: Halting Problem",
    fontsize=13,
    weight="bold"
)

plt.show()


# =====================================================================
# 10. GENERATED TEST SWEEP
# =====================================================================

H("10. Generated test sweep")

M("""
The next experiment generates arithmetic assignment programs with different
values. It compares the reference interpreter and compiler.

The sweep includes expressions that activate the injected faulty rewrite.
It is an empirical finite test, not a proof over all programs.
""")


generated = []

for a in range(0, 20):

    for b in range(0, 20):

        src = f"x = {b}; y = {a} - x; print(y);"

        generated.append(
            (f"generated_{a}_{b}", src)
        )


generated_correct = [
    execute_case(n, s, False)
    for n, s in generated
]

generated_faulty = [
    execute_case(n, s, True)
    for n, s in generated
]


sweep_correct_pass = sum(
    r["status"] == "PASS"
    for r in generated_correct
)

sweep_faulty_detected = sum(
    r["status"] == "BUG DETECTED"
    for r in generated_faulty
)

sweep_faulty_pass = sum(
    r["status"] == "PASS"
    for r in generated_faulty
)


sweep_df = pd.DataFrame(
    [
        [
            "Correct compiler",
            len(generated),
            sweep_correct_pass,
            len(generated) - sweep_correct_pass
        ],
        [
            "Injected-bug compiler",
            len(generated),
            sweep_faulty_pass,
            sweep_faulty_detected
        ]
    ],
    columns=[
        "Compiler mode",
        "Programs tested",
        "Matching outputs",
        "Mismatches detected"
    ]
)

display(sweep_df)


# =====================================================================
# 11. FINAL RESULTS AND CHARTS
# =====================================================================

H("11. Final benchmark results")

all_rows = []

for mode, result_list in [
    ("Correct compiler", correct_results),
    ("Injected-bug compiler", faulty_results)
]:

    for r in result_list:

        all_rows.append({

            "Compiler mode": mode,

            "Program": r["name"],

            "Reference output": r["reference_output"],

            "Compiled output": r["compiled_output"],

            "Result": r["status"],

            "Error": r["error"]

        })


final_report = pd.DataFrame(all_rows)

display(final_report)


correct_pass = sum(
    r["status"] == "PASS"
    for r in correct_results
)

correct_mismatch = sum(
    r["status"] == "BUG DETECTED"
    for r in correct_results
)

faulty_mismatch = sum(
    r["status"] == "BUG DETECTED"
    for r in faulty_results
)

faulty_pass = sum(
    r["status"] == "PASS"
    for r in faulty_results
)

errors = sum(
    r["status"] == "ERROR"
    for r in correct_results + faulty_results
)

unknowns = sum(
    r["status"] == "UNKNOWN / STEP LIMIT"
    for r in correct_results + faulty_results
)


summary = pd.DataFrame(
    [
        ["Curated benchmark programs", len(BENCHMARK)],
        ["Correct compiler: matching outputs", correct_pass],
        ["Correct compiler: mismatches", correct_mismatch],
        ["Faulty compiler: mismatches detected", faulty_mismatch],
        ["Faulty compiler: matching outputs", faulty_pass],
        ["Errors", errors],
        ["Unknown / step limit", unknowns],
        ["Generated programs in sweep", len(generated)],
        ["Generated correct-compiler matches", sweep_correct_pass],
        ["Generated faulty-compiler mismatches", sweep_faulty_detected]
    ],
    columns=["Metric", "Count"]
)

display(summary)


fig, ax = plt.subplots(figsize=(9, 4.5))

ax.bar(
    [
        "Correct compiler\nPASS",
        "Faulty compiler\nBUG DETECTED",
        "Faulty compiler\nPASS"
    ],
    [
        correct_pass,
        faulty_mismatch,
        faulty_pass
    ]
)

ax.set_ylabel("Number of benchmark programs")

ax.set_title(
    "Compiler Differential Testing Results"
)

plt.tight_layout()

plt.show()


fig, ax = plt.subplots(figsize=(8, 4))

ax.bar(
    [
        "Correct compiler\nmatching",
        "Faulty compiler\nmismatches"
    ],
    [
        sweep_correct_pass,
        sweep_faulty_detected
    ]
)

ax.set_ylabel("Number of generated programs")

ax.set_title(
    "Generated Test Sweep Results"
)

plt.tight_layout()

plt.show()


# =====================================================================
# 12. PRACTICAL ENGINEERING WORKAROUND
# =====================================================================

H("12. Practical engineering workaround")

M("""
A practical compiler team cannot rely on a universal exact correctness
decider. It can combine:

1. Differential testing against an independent interpreter/compiler.
2. Regression testing: preserve every discovered counterexample.
3. Randomized and property-based test generation.
4. Boundary-value testing and malformed-input testing.
5. Static analysis of intermediate representations.
6. Formal verification of selected compiler passes under explicit assumptions.
7. Time and resource limits, reported as operational limits rather than proof
   that a program loops forever.

The key principle is to report a concrete mismatch when one is observed and
avoid claiming universal correctness merely because a finite test suite passes.
""")


# =====================================================================
# 13. TEAM CONTRIBUTION STATEMENT
# =====================================================================

H("13. Team contribution statement")

M("""
| Group Member | Register Number | Suggested Contribution |
|---|---|---|
| PARIMI GOWTHAM | RA2411026050033 | Formal model, computability and undecidability analysis |
| DANDE DURGA PRASAD | RA2411026050066 | MiniLang parser, compiler and virtual-machine implementation |
| M. SHANMUKHA CHOWDHARY | RA2411026050064 | Test cases, validation, benchmark analysis and diagrams |

The contribution descriptions are suggested allocations. Edit them to reflect
the work each member actually completed. All members should review the final
results and report.
""")


# =====================================================================
# 14. EXPORT CSV REPORTS
# =====================================================================

final_report.to_csv(
    "Task4_Compiler_Bug_Detection_Benchmark.csv",
    index=False
)

summary.to_csv(
    "Task4_Compiler_Bug_Detection_Summary.csv",
    index=False
)

sweep_df.to_csv(
    "Task4_Compiler_Bug_Detection_Generated_Sweep.csv",
    index=False
)


pd.DataFrame([{

    "Program": r["name"],

    "Source": r["source"],

    "Bytecode": str(r["bytecode"]),

    "VM trace": " | ".join(r["vm_trace"]),

    "Reference output": r["reference_output"],

    "Faulty output": r["compiled_output"],

    "Detection": r["status"]

} for r in faulty_results]).to_csv(
    "Task4_Compiler_Bug_Detection_Bytecode_Trace.csv",
    index=False
)


H("14. Generated files")

M("""
The notebook creates these downloadable CSV reports in the Colab Files panel:

1. Task4_Compiler_Bug_Detection_Benchmark.csv
2. Task4_Compiler_Bug_Detection_Summary.csv
3. Task4_Compiler_Bug_Detection_Generated_Sweep.csv
4. Task4_Compiler_Bug_Detection_Bytecode_Trace.csv
""")


# =====================================================================
# 15. CONCLUSION
# =====================================================================

H("15. Conclusion")

M(f"""
This case study implements a restricted compiler-testing environment.
MiniLang programs are tokenized and parsed into an abstract syntax tree.
The reference interpreter executes the original AST, while the compiler
translates an optimized AST into virtual-machine bytecode. Differential
testing compares observable outputs and variable states.

A deliberately incorrect optimization (`0 - x` rewritten as `x`) is detected
by test programs. In the curated benchmark, the correct compiler has
**{correct_pass} matching cases out of {len(BENCHMARK)}**. The injected-bug
compiler produces **{faulty_mismatch} detected mismatches**. The generated
sweep tests **{len(generated)} programs**, finding **{sweep_faulty_detected}**
mismatches in the faulty mode.

These finite results demonstrate bug detection, not universal proof.
General program equivalence and universal compiler correctness are
undecidable for unrestricted programs. Practical compiler engineering
therefore combines testing, static analysis and formal verification of
restricted components.

**Group Members:**

1. PARIMI GOWTHAM — RA2411026050033
2. DANDE DURGA PRASAD — RA2411026050066
3. M. SHANMUKHA CHOWDHARY — RA2411026050064

**Course:** 21CSC301T – Formal Language and Automata
""")


print("=" * 72)
print("TASK 4: COMPILER BUG DETECTION - EXECUTION COMPLETED")
print("Group Members:")
print("1. PARIMI GOWTHAM - RA2411026050033")
print("2. DANDE DURGA PRASAD - RA2411026050066")
print("3. M. SHANMUKHA CHOWDHARY - RA2411026050064")
print("Curated benchmark:", len(BENCHMARK), "programs")
print("Correct compiler PASS:", correct_pass)
print("Injected compiler bugs detected:", faulty_mismatch)
print("Generated programs:", len(generated))
print("Generated faulty compiler mismatches:", sweep_faulty_detected)
print("CSV reports created: 4")
print("=" * 72)
````
